Array manipulation

stacking

you can't stack 2 arrays

```python
import numpy as np
```

```python
A=np.array([1,2,3,4])
B=np.array([5,6,7,8])
```

```python
A
```

```python
B
```

Vertical stacking

A and B are passed in form of tuple

```python
np.vstack((A,B))
```

horizontal stack

```python
np.hstack((A,B))
```

column stack

```python
np.column_stack((A,B))
```

Splitting arrays

```python
arr=np.arange(16).reshape(4,4)
arr
```

np.hsplit(array , no. of splits)-->hsplit splits array horizontally

```python
np.hsplit(arr,2)
```

vsplit() -->splits array horizontally

```python
arr=np.vsplit(arr,4)
```

```python
for i in arr:
    print(i)
```
